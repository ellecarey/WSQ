# Dictionaries
A key-value pair mapping

```python
# creating a dict
returns = {'AAPL':0.01,'XOM':-0.02,'BAC':0.05} 
print (returns)
print (type(returns))
```

```python
# Note: keys are typically strings, dates, or numbers values can be numbers, lists, strings, dictionaries, etc. 
mixd_dict={'a':'b',1:'c','list':[1,2,3,45],'dict':{'key':'value'}}
print (mixd_dict)

# 'a': 'b' — key is the string 'a', value is the string 'b'.
# 1: 'c' — key is the integer 1, value is the string 'c'.
# 'list': [1, 2, 3, 45] — key is the string 'list', value is a list of integers.
# 'dict': {'key': 'value'} — key is the string 'dict', value is another dictionary (a nested one).
```

```python
# accessing items
print (returns['XOM'])
```

```python
print(returns.values())
```

```python
print(returns.keys())
```

```python
# changing items
returns['AAPL']=-0.01
print (returns)
```

```python
# add items
returns['TWTR']=0.005
print (returns)
```

```python
# delete items
returns.pop('TWTR')
print (returns)
```

```python
returns.update({'XOM':-0.03,'BAC':0.04,'V':0.015})
print (returns)
```
