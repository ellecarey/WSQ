# Variables

```python
# variable assignment using '='
appl_ret = 0.01
market_ret = 0.02
```

```python
# print statement outputs variable values
print(appl_ret)
print(market_ret)
```

```python
# variables can be defined in terms of other variables
excess_ret = appl_ret - market_ret
print(excess_ret)
```
